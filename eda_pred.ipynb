```python
import pandas as pd 

df = pd.read_csv('matri_con_prediccion.csv')
```

```python
riesgo_total = df['riesgo_violencia'].sum()
total = len(df)
porcentaje = (riesgo_total / total) * 100

print(f"{porcentaje:.2f}% de los matrimonios fueron clasificados como de riesgo.")
```

```python
# Edad promedio en casos con y sin riesgo
df.groupby('riesgo_violencia')[['EDADHOM', 'EDADMUJ']].mean()
```

```python
# Riesgo por departamento
df.groupby('DEPREG')['riesgo_violencia'].mean().sort_values(ascending=False).head(5)
```

```python
import matplotlib.pyplot as plt

df['riesgo_violencia'].value_counts().plot(kind='bar')
plt.title("Clasificación de riesgo en matrimonios")
plt.xticks(ticks=[0,1], labels=['Sin riesgo', 'Con riesgo'], rotation=0)
plt.ylabel("Cantidad de matrimonios")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

sns.histplot(data=df, x='EDADMUJ', hue='riesgo_violencia', bins=20, kde=True)
plt.title("Distribución de edad de mujeres según riesgo")
plt.xlabel("Edad de la mujer")
plt.ylabel("Cantidad")
plt.legend(title="Riesgo", labels=["Sin riesgo", "Con riesgo"])
plt.tight_layout()
plt.show()
```
