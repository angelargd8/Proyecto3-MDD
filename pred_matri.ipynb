## Prediccion en modelo 

```python
import joblib
import pandas as pd 


# Cargar dataset de matrimonios
df_mapeado = pd.read_csv('df_mapeado.csv')
df_matri = pd.read_csv('matri_unite_data.csv')


# MODELO 
modelo = joblib.load('modelo_violencia_relog.pkl')

# Hacer predicciones
predicciones = modelo.predict(df_mapeado)

# Agregar al DataFrame original
df_matri['riesgo_violencia'] = predicciones

# Guardar resultados
df_matri.to_csv('matri_con_prediccion.csv', index=False)
```
