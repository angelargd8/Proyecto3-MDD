## Variables de matrimonio

- Departamento de registro (DEPREG): departamento donde se oficializó el matrimonio
- Minicipio de registro (MUPREG): municipio donde se oficializó el matrimonio
- Mes de registro (MESREG): mes en el que se oficializó el matrimonio
- Año de registro (AÑOREG): en el que se oficializó el matrimonio
- Clase de unión (CLAUNI): tipo de relación legal o convencional entre los miembros de la pareja
- Edad del hombre (EDADHOM): años de vida del hombre
- Edad de la mujer (EDADMUJ): años de vida de la mujer
- Grupo étnico del hombre (GETHOM): etnia del hombre. Solo especifica si es indigena o no o si no se conoce el dato
- Grupo étnico de la mujer (GETMUJ): etnia de la mujer. Solo especifica si es indigena o no o si no se conoce el dato
- Nacionalidad del hombre (NACHOM): país de nacionalidad del hombre
- Nacionalidad de la mujer (NACMUJ): país de nacionalidad de la mujer
- Ocupación del hombre (OCUHOM): trabajo del hombre
- Ocupación de la mujer (OCUMUJ): trabajo de la mujer
- Número de nupcias contraidas del hombre (NUPHON): cantidad de veces que el hombre se ha casado
- Número de nupcias contraidas de la mujer (NUPMUJ): cantidad de veces que la mujer se ha casado
- Departamento de ocurrencia (DEPOCU): Departamento en donde se realizó el matrimonio
- Municipio de ocurrencia (MUPOCU): Municipio en donde se realizó el matrimonio
- Mes de ocurrencia (MESOCU): mes en el que se realizó el matrimonio
- Año de ocurrencia (AÑOOCU): año en el que se realizó el matrimonio
- Área geográfica de ocurrencia (AREAG): si ocurrió en un lugar urbano o rural

Calculamos los datos "Ignorado" que hay en cada variable numérica. Como son menos del 5% de los datos, lo trataremos como valores nulos. 

```python
# Llenar datos vacios
cols_a_imputar = ["EDADHOM", "EDADMUJ", "NUPHON", "NUPMUJ"]
df[cols_a_imputar] = df[cols_a_imputar].replace("Ignorado", np.nan)
df[cols_a_imputar] = df[cols_a_imputar].apply(pd.to_numeric, errors="coerce")

df.fillna({
    "MUPREG": "Ignorado",
    "MESREG": "Ignorado",
    "CLAUNI": "Ignorado",
    "GETHOM": "Ignorado",
    "GETMUJ": "Ignorado",
    "DEPREG": "Ignorado",
    "NACHOM": "Ignorado",
    "NACMUJ": "Ignorado",
    "OCUHOM": "Ignorado",
    "OCUMUJ": "Ignorado",
    "DEPOCU": "Ignorado",
    "MUPOCU": "Ignorado",
    "MESOCU": "Ignorado",
    "AREAG": "Ignorado"
}, inplace=True)

df.fillna({
    "AÑOREG": df["AÑOREG"].median(),
    "EDADHOM": df["EDADHOM"].median(), 
    "EDADMUJ": df["EDADMUJ"].median(),
    "NUPHON": df["NUPHON"].median(),
    "NUPMUJ": df["NUPMUJ"].median(), 
    "AÑOOCU": df["AÑOOCU"].median()
}, inplace=True)

print("--------- INFORMACIÓN ---------")
df.info()

print("\n--------- DATOS NULOS ---------")
print(df.isnull().sum())
```

Convertimos los datos de las variables numéricas para que se acepten como nums. Se puede observar que ahora ya no existen valores nulos y las variables que deberían ser numéricas ya lo son.

## Determinar la tendencia al agrupamiento y el mejor número de “clusters” a utilizar

```python
file_path = "matrimonios.csv"
df = pd.read_csv(file_path)

# Variables relevantes para clustering
df_clustering = df[['EDADHOM', 'EDADMUJ', 'AÑOREG']].copy()

# Convertir a numérico y eliminar valores no válidos
df_clustering = df_clustering.apply(pd.to_numeric, errors='coerce')
df_clustering.dropna(inplace=True)

# Estandarizar los datos
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_clustering)
```

### Método del Codo para determinar el mejor número de clusters 


```python
wcss = []  # Within-cluster sum of squares
K_range = range(2, 10)  # Probamos de 2 a 10 clusters

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_scaled)
    wcss.append(kmeans.inertia_)

# Graficar el método del codo
plt.figure(figsize=(8, 5))
plt.plot(K_range, wcss, marker='o', linestyle='--')
plt.xlabel('Número de Clusters')
plt.ylabel('WCSS (Suma de Cuadrados Dentro del Cluster)')
plt.title('Método del Codo para Determinar el Número Óptimo de Clusters')
plt.grid(True)
plt.show()
```

Para seleccionar el número adecuado de clusters, se utilizó el **Método del Codo**, que evalúa la suma de los cuadrados dentro de cada cluster (**WCSS - Within-Cluster Sum of Squares**). En el gráfico generado, se observa que la curva tiene una caída pronunciada hasta alrededor de **k=3 o k=4**, y luego la disminución es menos pronunciada. 

Este comportamiento indica que **3 o 4 clusters** podrían ser una buena elección, ya que después de este punto, agregar más clusters no reduce significativamente la variabilidad dentro de los grupos. 

Con base en esta observación, seleccionaremos **k=3 o k=4** y verificaremos la calidad del agrupamiento mediante el **coeficiente de silueta** en el siguiente análisis.


```python
print(df_scaled.shape)
```

### Evaluar calidad del clustering con la Silueta


```python

df_sample = df_scaled[np.random.choice(df_scaled.shape[0], 50000, replace=False)]


silhouette_scores = []
for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(df_sample)
    silhouette_scores.append(silhouette_score(df_sample, cluster_labels))

# Graficar la puntuación de silueta
plt.figure(figsize=(8, 5))
plt.plot(K_range, silhouette_scores, marker='o', linestyle='--', color='red')
plt.xlabel('Número de Clusters')
plt.ylabel('Puntuación de Silueta')
plt.title('Coeficiente de Silueta para Diferentes Clusters')
plt.grid(True)
plt.show()

# Aplicar K-Means con el mejor número de clusters
best_k = K_range[np.argmax(silhouette_scores)]  # Elegimos el k con mejor silueta
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df_clustering['Cluster'] = kmeans.fit_predict(df_scaled)
```

Para evaluar la calidad del agrupamiento, se utilizó el **Coeficiente de Silueta**, que mide qué tan bien separados están los clusters y qué tan distintos son entre sí. Un valor alto indica que los clusters están bien definidos.

En el gráfico generado, se observa que el coeficiente de silueta es más alto para **k=2** (aproximadamente **0.52**), pero como el método del codo sugirió **k=3 o k=4**, también es importante revisar estos valores. 

Para **k=3**, la puntuación de silueta es alrededor de **0.39**, y para **k=4** es ligeramente menor pero aún aceptable. A medida que el número de clusters aumenta, la calidad del agrupamiento disminuye.

Basado en ambos métodos:
- **El método del codo** sugirió que **k=3 o k=4** es un buen punto de agrupamiento.
- **El coeficiente de silueta** muestra que **k=3 tiene una mejor puntuación** en comparación con k=4.
- Se elige **k=3 como el número óptimo de clusters** para continuar con el análisis.



### Interpretación de los Clusters


```python
# Eliminando valores como "Ignorado"
df['EDADHOM'] = pd.to_numeric(df['EDADHOM'], errors='coerce')
df['EDADMUJ'] = pd.to_numeric(df['EDADMUJ'], errors='coerce')

# Filtrar edades para evitar outliers
df_filtered = df[(df['EDADHOM'] >= 10) & (df['EDADHOM'] <= 100)]
df_filtered = df_filtered[(df_filtered['EDADMUJ'] >= 10) & (df_filtered['EDADMUJ'] <= 100)]

plt.figure(figsize=(10, 6))
sns.boxplot(x=df_filtered['Cluster'], y=df_filtered['EDADHOM'])
plt.xlabel('Cluster')
plt.ylabel('Edad del Hombre')
plt.title('Distribución de Edad del Hombre por Cluster')
plt.ylim(10, 100)  
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
sns.boxplot(x=df_filtered['Cluster'], y=df_filtered['EDADMUJ'])
plt.xlabel('Cluster')
plt.ylabel('Edad de la Mujer')
plt.title('Distribución de Edad de la Mujer por Cluster')
plt.ylim(10, 100)
plt.grid(True)
plt.show()


plt.figure(figsize=(10, 6))
sns.boxplot(x=df['Cluster'], y=df['AÑOREG'])
plt.xlabel('Cluster')
plt.ylabel('Año del Matrimonio')
plt.title('Distribución de Años de Matrimonio por Cluster')
plt.grid(True)
plt.show()
```

Para analizar los grupos identificados en el clustering, se han visualizado las distribuciones de las variables **Edad del Hombre**, **Edad de la Mujer** y **Año del Matrimonio** por cluster.

### **Distribución de Edad de la Mujer por Cluster**
- **Cluster 0**: Similar a la distribución de los hombres, las mujeres en este grupo tienen una mediana de **40 a 50 años**, con edades que oscilan entre los **15 y 80 años**.

Estos resultados sugieren que los **clusters reflejan dos perfiles distintos de matrimonios**:
1. **Matrimonios con parejas de mayor edad (Cluster 0)**.

###  **Distribución del Año del Matrimonio por Cluster**
El gráfico de distribución de los años de matrimonio no muestra diferencias significativas entre los clusters. Ambos grupos tienen matrimonios distribuidos entre **2010 y 2022**, lo que indica que la tendencia de los matrimonios por edad se ha mantenido constante a lo largo del tiempo.

El análisis revela que la segmentación del clustering se basa **principalmente en la edad de los contrayentes**. Se han identificado dos grupos distintos:
- **Cluster 0**: Matrimonios entre personas mayores.
- **Cluster 1**: Matrimonios entre personas más jóvenes.
