## Variables de matrimonio

- Departamento de registro (DEPREG): departamento donde se oficializó el matrimonio
- Minicipio de registro (MUPREG): municipio donde se oficializó el matrimonio
- Mes de registro (MESREG): mes en el que se oficializó el matrimonio
- Año de registro (AÑOREG): en el que se oficializó el matrimonio
- Clase de unión (CLAUNI): tipo de relación legal o convencional entre los miembros de la pareja
- Edad del hombre (EDADHOM): años de vida del hombre
- Edad de la mujer (EDADMUJ): años de vida de la mujer
- Grupo étnico del hombre (GETHOM): etnia del hombre. Solo especifica si es indigena o no o si no se conoce el dato
- Grupo étnico de la mujer (GETMUJ): etnia de la mujer. Solo especifica si es indigena o no o si no se conoce el dato
- Nacionalidad del hombre (NACHOM): país de nacionalidad del hombre
- Nacionalidad de la mujer (NACMUJ): país de nacionalidad de la mujer
- Ocupación del hombre (OCUHOM): trabajo del hombre
- Ocupación de la mujer (OCUMUJ): trabajo de la mujer
- Número de nupcias contraidas del hombre (NUPHON): cantidad de veces que el hombre se ha casado
- Número de nupcias contraidas de la mujer (NUPMUJ): cantidad de veces que la mujer se ha casado
- Departamento de ocurrencia (DEPOCU): Departamento en donde se realizó el matrimonio
- Municipio de ocurrencia (MUPOCU): Municipio en donde se realizó el matrimonio
- Mes de ocurrencia (MESOCU): mes en el que se realizó el matrimonio
- Año de ocurrencia (AÑOOCU): año en el que se realizó el matrimonio
- Área geográfica de ocurrencia (AREAG): si ocurrió en un lugar urbano o rural

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Preprocesamiento y limpieza

```python
# Cargar el dataset (ajusta la ruta según corresponda)
df = pd.read_csv("matrimonios.csv") 
columnas_validas = ['DEPREG','MUPREG','MESREG','AÑOREG','CLAUNI','EDADHOM','EDADMUJ','GETHOM','GETMUJ','NACHOM','NACMUJ','OCUHOM','OCUMUJ','NUPHON','NUPMUJ','DEPOCU','MUPOCU','MESOCU','AÑOOCU','AREAG']
df = df[columnas_validas]

# Mostrar las primeras filas para inspección
df.head()
```

Se eliminaron las otras columnas porque están en su mayoría vacias y no aportan datos relevantes.

```python
# Información general del dataset
print("--------- INFORMACIÓN ---------")
df.info()

# Verificar valores únicos por columna (especialmente en categóricas)
print("\n--------- DATOS ÚNICOS ---------")
df.nunique()
```

Aquí podemos ver que datos que deberían ser numéricos, como las edades aparecen como objetos, lo que significa que los datos no están guardados de manera correcta o existen datos categoricos dentro de los numéricos. Además, los datos de etnia solo deberían ser 3 pero son 4, lo que significa que hay un dato extra. 

```python
# Contar valores nulos por columna
print("--------- DATOS NULOS ---------")
print(df.isnull().sum())

print("\n--------- FORMA DEL DATASET ---------")
print(df.shape)
```

Aquí podemos ver la cantidad de datos nulos que hay en cada variable. También se puede ver la forma del dataset para comparar con los nulos.

```python
# Contar cuántas veces aparece "Ignorado" en cada columna
num_cols = ['AÑOREG','EDADHOM','EDADMUJ','NUPHON','NUPMUJ','AÑOOCU']
df_num = df[num_cols]
ignorado_counts = (df_num == "Ignorado").sum()

# Mostrar solo las columnas donde hay valores "Ignorado"
ignorado_counts
```

Calculamos los datos "Ignorado" que hay en cada variable numérica. Como son menos del 5% de los datos, lo trataremos como valores nulos. 

```python
# Llenar datos vacios
cols_a_imputar = ["EDADHOM", "EDADMUJ", "NUPHON", "NUPMUJ"]
df[cols_a_imputar] = df[cols_a_imputar].replace("Ignorado", np.nan)
df[cols_a_imputar] = df[cols_a_imputar].apply(pd.to_numeric, errors="coerce")

df.fillna({
    "MUPREG": "Ignorado",
    "MESREG": "Ignorado",
    "CLAUNI": "Ignorado",
    "GETHOM": "Ignorado",
    "GETMUJ": "Ignorado",
    "DEPREG": "Ignorado",
    "NACHOM": "Ignorado",
    "NACMUJ": "Ignorado",
    "OCUHOM": "Ignorado",
    "OCUMUJ": "Ignorado",
    "DEPOCU": "Ignorado",
    "MUPOCU": "Ignorado",
    "MESOCU": "Ignorado",
    "AREAG": "Ignorado"
}, inplace=True)

df.fillna({
    "AÑOREG": df["AÑOREG"].median(),
    "EDADHOM": df["EDADHOM"].median(), 
    "EDADMUJ": df["EDADMUJ"].median(),
    "NUPHON": df["NUPHON"].median(),
    "NUPMUJ": df["NUPMUJ"].median(), 
    "AÑOOCU": df["AÑOOCU"].median()
}, inplace=True)

print("--------- INFORMACIÓN ---------")
df.info()

print("\n--------- DATOS NULOS ---------")
print(df.isnull().sum())
```

Convertimos los datos de las variables numéricas para que se acepten como nums. Para las variables categóricas reemplazamos los valores nulos con "Ignorados" y en las variables numéricas los reemplazamos por la media. Se puede observar que ahora ya no existen valores nulos y las variables que deberían ser numéricas ya lo son. 

```python
# Convertir todo a minúsculas y quitar espacios extra
df["GETHOM"] = df["GETHOM"].str.strip().str.lower()
df["GETMUJ"] = df["GETMUJ"].str.strip().str.lower()

print(df["GETHOM"].unique())
print(df["GETMUJ"].unique())


```

Arreglamos los valores de etnia tanto en hombres como mujeres, para que muestren solo los valores que realmente son. Para esto eliminamos espacios y colocamos todos en minúscula.

```python
# Edades de matrimonio

# Estadísticas descriptivas
print(df[['EDADHOM', 'EDADMUJ']].describe())

# Histogramas
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(df['EDADHOM'], bins=30, kde=True, ax=ax[0])
ax[0].set_title("Distribución de Edad del Hombre")

sns.histplot(df['EDADMUJ'], bins=30, kde=True, ax=ax[1])
ax[1].set_title("Distribución de Edad de la Mujer")

plt.show()

# Diagramas de caja para ver outliers
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(y=df['EDADHOM'], ax=ax[0])
ax[0].set_title("Boxplot de Edad del Hombre")

sns.boxplot(y=df['EDADMUJ'], ax=ax[1])
ax[1].set_title("Boxplot de Edad de la Mujer")

plt.show()
```

En esta sección podemos ver las edades en las que las personas, tanto hombres como mujeres, se casan. El análisis estadístico muestra que la edad promedio para casarse es de 28 año para los hombres y de 26 para las mujeres. Además, el hombre más joven en casarse tenia 14 años y el más grande 99. Por otro lado, la mujer más joven tenía 12 años y la más grande tenía 109. Los histogramas validan esta información y, también, muestran que la edad de los matrimonios empieza a disminuir drásticamente después de los 30 años para ambos sexos. Otra cosa importante es que las mujeres suelen casarse más jovenes que los hombres. Esto también se puede ver en las gráficas de caja y bigotes. Estas últimas gráficas también muestran que las edades menos comunes, es decir los datos atípicos, se encuentran por encima de los 45 para los hombres y encima de los 39 para las mujeres. En su mayoría, los datos atípicos se encuentran en edades elevadas, pero también existen edades bajas, como se vio en el análisis estadístico.

```python
# Diferencias de edad
df["DIF_EDAD"] = df["EDADHOM"] - df["EDADMUJ"]

plt.figure(figsize=(12,5))
sns.countplot(x=df["DIF_EDAD"], color="purple", order=sorted(df["DIF_EDAD"].unique()))
plt.xticks(rotation=90)
plt.title("Diferencia de Edad entre los Cónyuges")
plt.xlabel("Diferencia de Edad (Hombre - Mujer)")
plt.ylabel("Cantidad de Registros")
plt.show()
```

Aquí podemos ver como varia la edad entre las parejas del matrimonio. Los números positivos significan que el hombre tiene mayor edad, mientras que los negativos que es la mujer la más grande. Lo más común, por mucho, es que el hombre sea 2 años mayor que su pareja. Luego es que sea 1 mayor y como tercer lugar es que sea 3 años mayor. Por lo tanto diríamos que lo más común es que en una pareja el hombre sea más grande que la mujer. Sin embargo, podemos ver que es del lado de los negativos, es decir una mujer de más edad, donde podemos ver la diferencia más notoria de años, siendo esta de 89.

```python
cat_vars = ["CLAUNI", "GETHOM", "GETMUJ", "NACHOM", "NACMUJ", "OCUHOM", "OCUMUJ", "AREAG"]

# Tablas de frecuencia y proporciones
for col in cat_vars:
    print(f"\nFrecuencia de {col}:")
    print(df[col].value_counts(normalize=True) * 100)

# Gráficos de barra
fig, axes = plt.subplots(4, 2, figsize=(15, 20))

for i, col in enumerate(cat_vars):
    if col in ['NACHOM', 'NACMUJ', 'OCUHOM', 'OCUMUJ']:  
        # Solo tomar las 5 categorías más frecuentes para 'NACHOM', 'NACMUJ', 'OCUHOM', 'OCUMUJ'
        top_5 = df[col].value_counts().nlargest(5).index
        sns.countplot(y=df[col], order=top_5, ax=axes[i//2, i%2])
        axes[i//2, i%2].set_title(f"Top 5 Distribución de {col}")
    else:
        sns.countplot(y=df[col], order=df[col].value_counts().index, ax=axes[i//2, i%2])
        axes[i//2, i%2].set_title(f"Distribución de {col}")

plt.tight_layout()
plt.show()


```

Aquí podemos observar los datos más relevantes en las variables categóricas. Lo primero que se ve es que la unión más común es la de "comunidad de gananciales", que es cuando se regula la distribución de los bienes que adquiere cada mitad de la pareja. También se ve que tanto para hombres, como para mujeres, los datos de etnia se desconocen en su gran mayoría. Las nacionalidades tambien son un dato bastante equilibrado para hombres y mujeres, teniendo la gran mayoría de datos en Guatemala con un 83%. Luego están las frecuencias para las ocupaciones. Aquí podemos ver como el dato mayor es desconocido en ambos géneros con un 85%. Pero luego tenemos podemos ver una difencia clave entre la ocupación de los hombres y las mujeres, ya que los primeros tienden a trabajar más en áreas agricultoras, mientras que las segundas suelen quedarse como amas de casa. Por último podemos ver como el área geográfica también tiene el valor desconocido como el más alto por mucho más que los otros dos datos con un 92% de la data recolectada.

Después, podemos ver en las gráficas los mismos resultados que ya se mencionaron anteriormente. Con esto podemos confirmar que las variables de nacionalidad, tanto de hombre como mujer, y clase de unión serán las únicas que aportaran datos verdaderamente relevantes al momento de hacer el análisis, ya que se conoce la mayoría de los datos.

```python
plt.figure(figsize=(12, 6))
sns.countplot(y=df['DEPREG'], order=df['DEPREG'].value_counts().index)
plt.title("Matrimonios por Departamento de Registro")
plt.show()

plt.figure(figsize=(12, 6))
sns.countplot(y=df['DEPOCU'], order=df['DEPOCU'].value_counts().index)
plt.title("Matrimonios por Departamento de Ocurrencia")
plt.show()
```

En estas gráficas se puede ver de forma clara como se "enfrentan" los datos de el lugar donde se oficializó un matrimonio y el lugar donde este ocurrió. Vemos que aquí, al igual que en muchas de las variables categóricas, desconocer los datos es lo más común. Sin embargo, en este caso si podemos sacar información importante, que nos servirá al momento de predecir la violencia. Esto es el hecho de que que estos datos son bastante igualados, lo que nos confirma que lo más común es que las personas se mantengan en el mismo lugar donde se oficializa el matrimonio.

```python
# Gráfico de dispersión entre edades
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['EDADHOM'], y=df['EDADMUJ'], alpha=0.5)
plt.xlabel("Edad del Hombre")
plt.ylabel("Edad de la Mujer")
plt.title("Relación entre Edades")
plt.show()
```

```python
# Matriz de correlación
plt.figure(figsize=(8, 6))
sns.heatmap(df[['EDADHOM', 'EDADMUJ']].corr(), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Matriz de Correlación")
plt.show()
```

Continuando con el análisis de la edad, en estas dos gráficas podemos reforzar lo que se había hablado antes acerca de la relación de las edades. En la primera gráfica podemos ver que las edades se inclinan más al lado de los hombres, confirmando que son estos los que suelen tener una mayor edad. Algo curioso es que hay una edad en los hombres después de los 25, donde parece que tienden a casarse con mujeres más grandes, ya que se crea una línea de puntos de forma vertical (es decir cambiando la edad de las mujeres pero no de los hombres) en mujeres arriba de los 60 años. La segunda gráfica muestra la relación entre las edades en forma de matriz, demostrando que en un 85% de los casos, las personas suelen casarse con personas en el mismo rango de edad. 

```python
low = df[['EDADHOM', 'EDADMUJ']].quantile(0.01)  # 1% más bajo
high = df[['EDADHOM', 'EDADMUJ']].quantile(0.99)  # 1% más alto

df_filtered = df[
    (df['EDADHOM'] >= low['EDADHOM']) & (df['EDADHOM'] <= high['EDADHOM']) &
    (df['EDADMUJ'] >= low['EDADMUJ']) & (df['EDADMUJ'] <= high['EDADMUJ'])
]

df_filtered.shape
```

```python
# Gráfico de dispersión entre edades
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df_filtered['EDADHOM'], y=df_filtered['EDADMUJ'], alpha=0.5)
plt.xlabel("Edad del Hombre")
plt.ylabel("Edad de la Mujer")
plt.title("Relación entre Edades")
plt.show()
```

Eliminando los outliers (datos atípicos) podemos ver que ahora tenemos una distribución casi completamente cuadrada. La diferencia es que la edad de matrimonio de los hombres llega a los 70 años mientras que las de las mujeres tiene más valores por debajo de los 20 que los hombres. También podemos observar como es más común que las mujeres menores de 20 se casen con hombres mayores de 60 a que hombres menores de 20 se casen con mujeres mayores de los 45.
